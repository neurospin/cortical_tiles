Prendre le crop, le transformer en mesh et le représenter ensemble avec le white mesh, en utilisant les conversions directes d'anatomist

```python
def remove_hull(arr, padding, ext):
    """Removes the pixels on the hull
    
    Pixels on the hull are defined as being in contact with both the internal and external part
    Parameters:
        arr: numpy array modified in the array
        padding: padding of the image, equal to the extension ext
        ext: local array extension in which to look for external and internal pixels"""
    arr_pad = np.pad(arr, ((padding,padding), (padding,padding), (padding,padding), (0,0)), 'constant', constant_values=0)
    EXTERNAL = 11 # Value of external part
    INTERNAL = 0 # Value of the internal part of the brain
    coords = np.argwhere(arr_pad > EXTERNAL)

    for i,j,k,l in coords:
        if arr_pad[i,j,k,l] != EXTERNAL and arr_pad[i,j,k,l] != INTERNAL:
            local_array = arr_pad[i-ext:i+ext+1,j-ext:j+ext+1,k-ext:k+ext+1,l]
            if np.any(local_array == 11):
                arr[i-padding,j-padding,k-padding,l] = 0

def threshold_and_binarize(arr, threshold):
    """threshold image"""
    arr[np.where(arr < threshold)]= 0
    arr[np.where(arr >= threshold)]= 32767

def mesh_without_hull(vol, padding, ext, threshold):
    arr = np.asarray(vol)

    # Removes hull
    remove_hull(arr, padding, ext)

    # Threshold and "binarize"
    threshold_and_binarize(arr, threshold)

    # Convert to bucket
    convert = aims.Converter_rc_ptr_Volume_S16_BucketMap_VOID()
    bck_map = convert(vol)
    bck = bck_map[0]

    # Displays bucket file
    mesh = cld.aims_tools.bucket_to_mesh(bck)
    return mesh
```
